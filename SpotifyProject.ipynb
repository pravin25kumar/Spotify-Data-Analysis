#### Q2 : Which primary music genres (primary_genre_l1) have the highest number of artists?

### Q4 : What percentage of all artists belongs to each primary genre?

```python
artist_count.head()
```

```python
artist_count['artist_pct'] = artist_count['artist_pct'].apply(lambda x : round(x, 2))
```

### Q5 : How has the number of artists changed based on their career start year?

```python
artists_by_year = df['career_start_year'].value_counts().sort_index().reset_index(name = 'no_of_artist').sort_values('no_of_artist', ascending = True)
```

```python
artists_by_year
```

```python
alt.Chart(artists_by_year).mark_line(strokeWidth=7,color='orange').encode(
    x= 'career_start_year:N',
    y= 'no_of_artist:Q',
    tooltip = ['career_start_year','no_of_artist']
).properties(
    width = 600,
    height= 400
)
```

#### Findings : 
- No. of artists in each year were below 100 till 1999. 
- There was a huge surge in the number of artists (5535) in the year 2000.
- After that it dipped down to 2355 in the year 2003 and ramained almost flat till 2016 (2633).
- in 2019 we can see the rise in number of artists (5375).
- Since then it has been in the decreasing path.

### Q8 : How has the popularity of different music genres changed across career-start periods?

```python
# music_genre_popularity = df.groupby(['career_start_year','primary_genre_l1']).size().reset_index(name="artists")
```

```python
music_genre_popularity = df.groupby(
    ['career_start_year','primary_genre_l1']
    )['primary_genre_l1'].size().reset_index(
        name="artists"
        ).sort_values(
            ['career_start_year','artists'], ascending = [True, False]
            ).drop_duplicates(
                'career_start_year'
                )
```

```python
music_genre_popularity
```

```python
alt.Chart(music_genre_popularity).mark_bar().encode(
    x = 'career_start_year:N',
    y = 'primary_genre_l1:N',
    tooltip = ['career_start_year', 'primary_genre_l1','artists'],
    color = 'primary_genre_l1:N'
).properties(
    width =650,
    height = 400
)
```

### Findings :
- Since 1957-1989, 'Rock' has been the mostly popular genre
- From 1990-1996, 'R&B' was mostly popular.
- From 2000-2022, 'Electronic' has been the most popular genre.

#### Q9 : Which combinations of primary_genre_l1 and primary_genre_l2 occur most frequently?

```python
freq_genre_combo = df.groupby(
    ['primary_genre_l1','primary_genre_l2']
    ).size().reset_index(name = 'occurence').sort_values(by= 'occurence', ascending = False).head(5)
```

```python
freq_genre_combo
```

```python
freq_genre_combo['Combo_genre'] = freq_genre_combo['primary_genre_l1'] + '+' + freq_genre_combo['primary_genre_l2']
```

```python
alt.Chart(freq_genre_combo).mark_bar().encode(
    y = alt.Y('Combo_genre:N', sort = alt.SortField('occurence', order = 'descending')),
    x = 'occurence:Q',
    color= 'Combo_genre:N',
    tooltip= ['primary_genre_l1','primary_genre_l2','occurence']
).properties(
    width = 500,
    height= 220
)
```

#### Based on the data :
- Primary Genre: Hip Hop & Secondary Genre: Hip Hop occurs most frequently.
- Primary Genre: Jazz & Secondary Genre: Jazz is at the 2nd position.
- Primary Genre: Electronic & Secondary Genre: EDM is at the 3rd position.

### Q12 : Among the 1.49% of artists with Spotify data, which genres have the highest average Spotify popularity and follower count?


```python
Genre_avg_popularity = df.dropna(
    subset= ['spotify_followers']
    ).groupby('primary_genre_l1')['spotify_followers'].mean().reset_index(name = 'avg_follower').sort_values('avg_follower', ascending = False)
```

```python
Genre_avg_popularity['avg_follower'] = Genre_avg_popularity['avg_follower'].round(0).astype('int64')
```

```python
Genre_avg_popularity
```

```python
alt.Chart(Genre_avg_popularity).mark_bar().encode(
    x= alt.X('primary_genre_l1:N', sort = alt.SortField(field = 'avg_follower', order = 'descending'), axis = alt.Axis(labelAngle = 45)),
    y = 'avg_follower:Q',
    color = 'avg_follower:Q',
    tooltip = ['primary_genre_l1', 'avg_follower']
).properties(
    width = 500,
    height = 300
)
```

### Findings : 
- Among all the genres, "Latin" category has the highest average follower count with 20950612 average followers.
- Hip Hop comes second with 9926661 average followers which is less than half the average count of Latin.
- Pop stands at third place with 9162600 average followers, close to Hip Hop.
